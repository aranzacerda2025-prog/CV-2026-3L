```python
import cv2 
from ultralytics import yolo

model = YOLO("yolov8n.pt")

cap = cv2.VideoCapture (0)
while True: 
ret, frame = cap.read()
if not ret:
   break
# El model realiza la prediccion directamente sobre el fotograma
results = model (frame)
annotated_frame = results [0].plot()
cv2.imshow("Reconocimiento" YOLOv8", annotated_frame)

if cv2.waitkey(1)& 0xFF == ord ("q"):
  break

cap.release()
cv2.destroyAllWINDOWS()


```
